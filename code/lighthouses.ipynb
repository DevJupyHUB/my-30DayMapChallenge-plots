```python
import geopandas as gpd
import pandas as pd
import matplotlib.pyplot as plt
import matplotlib.font_manager as fm
from matplotlib.animation import FuncAnimation
```

```python
lighthouses = gpd.read_file("lighthouses-japan.geojson")
```

```python
lighthouses.crs
```

```python
japan = gpd.read_file("gadm41_JPN_1.json")
```

```python
japan.crs
```

```python
# Register the font
font_path = "Lighthouse_PersonalUse.ttf"
fm.fontManager.addfont(font_path)
font_name = fm.FontProperties(fname=font_path).get_name()
```

```python
flicker = [True, False, True, False, False, True, False, True, True, False]

fig, ax = plt.subplots(figsize=(10, 10), facecolor="#0b1026")
fig.subplots_adjust(left=0.09, right=0.95)

# Static elements 
# Japan boundary
japan.plot(ax=ax, color="#1c2961", ec="#1c2961")

# Title
fig.text(0.5, 0.9, "Lighthouses of Japan", fontproperties=fm.FontProperties(fname=font_path),
         size=60, alpha=0.9, color="white", ha="center")

# Caption
fig.text(0.5, 0.02,
         "Lighthouse locations: GeoNames, CC BY 4.0, downloaded via QuickMapTools.\n"
          "Administrative boundaries: GADM 4.1 (Global Administrative Areas), https://gadm.org.\n"
          "Font: Lighthouse by Måns Grebäck. https://www.dafont.com/lighthouse2.font\n"
          "#30DayMapChallenge Day1 Points | Map: Bernadett Piros", 
         size=8, color="white", alpha=0.9, ha="center")

ax.axis("off")

# Lighthouses
lighthouses.plot(ax=ax, markersize=3, color="white")

# Last plotted points
point_artists = ax.collections[-1]

# Hide the points before ani start
point_artists.set_visible(False)

def update(frame):
    # True or False
    visible = flicker[frame % len(flicker)]
    # Show or hide 
    point_artists.set_visible(visible)    
    
# Animation
ani = FuncAnimation(fig, update, frames=len(flicker), interval=500, repeat=True, blit=False)

# Save
ani.save("lighthouses_japan.gif", writer="pillow", fps=2)

plt.close(fig)
```
